# Training Curves

Use this notebook for local inspection of synced TensorBoard scalar curves. The reusable parsing and plotting logic lives in `notebooks/utils/`; this notebook only selects runs and figures.

```python
from pathlib import Path

from utils.extract_training_curves import extract_training_curves
from utils.plot_training_curves import load_curves, plot_scalar

repo_root = Path.cwd()
if not (repo_root / "pyproject.toml").exists():
    repo_root = repo_root.parent

runs_root = repo_root / "outputs/finetuning"
curves_path = repo_root / "notebooks/analysis/training_curves.csv"
figures_dir = repo_root / "notebooks/analysis/figures"

# Latest completed matched diagnostic and full-run trios. Jobs 11286-11288
# crashed during their validation trace and therefore have no usable curves.
# Replace this trio after the corrected diagnostic completes.
initial_jobs = {"11197", "11198", "11199"}
full_jobs = {"11270", "11271", "11272"}
initial_job_label = "-".join(sorted(initial_jobs))
full_job_label = "-".join(sorted(full_jobs))

extract_training_curves(
    runs_root=runs_root,
    output=curves_path,
    jobs=initial_jobs | full_jobs,
)
curves = load_curves(curves_path)
```

```python
plot_scalar(
    df=curves,
    tag="val/loss",
    jobs=initial_jobs,
    output=figures_dir / f"initial_2b_{initial_job_label}_validation_loss.png",
    title="Earlier 2B initial convergence diagnostic",
    ylabel="Bounded validation loss",
)
```

```python
plot_scalar(
    df=curves,
    tag="train/loss_step",
    jobs=initial_jobs,
    output=figures_dir / f"initial_2b_{initial_job_label}_training_loss.png",
    title="Earlier 2B initial convergence diagnostic",
    ylabel="Training loss",
)
```

```python
plot_scalar(
    df=curves,
    tag="val/loss",
    jobs=full_jobs,
    output=figures_dir / f"full_2b_{full_job_label}_validation_loss.png",
    title="2B full finetuning runs",
    ylabel="Bounded validation loss",
)
```

```python
plot_scalar(
    df=curves,
    tag="train/loss_step",
    jobs=full_jobs,
    output=figures_dir / f"full_2b_{full_job_label}_training_loss.png",
    title="2B full finetuning runs",
    ylabel="Training loss",
)
```

```python
plot_scalar(
    df=curves,
    tag="train/loss_step",
    jobs=full_jobs,
    max_step=1_000,
    output=figures_dir / f"full_2b_{full_job_label}_training_loss_early.png",
    title="2B full finetuning: early learning",
    ylabel="Training loss",
)
```

```python
plot_scalar(
    df=curves,
    tag="train/loss_step",
    jobs=full_jobs,
    min_step=500,
    output=figures_dir / f"full_2b_{full_job_label}_training_loss_after_500_steps.png",
    title="2B full finetuning: after early learning",
    ylabel="Training loss",
)
```
