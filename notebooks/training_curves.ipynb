Use this notebook for local inspection of synced TensorBoard scalar curves. The reusable parsing and plotting logic lives in `notebooks/utils/`.

```python
from pathlib import Path

import pandas as pd

from utils.plot_training_curves import load_curves, plot_scalar

curves_path = Path("../outputs/analysis/training_curves.csv")
figures_dir = Path("../outputs/analysis/figures")
```

```python
jobs = {"11197", "11198", "11199"}
plot_scalar(
    df,
    tag="val/loss",
    jobs=jobs,
    output=figures_dir / "initial_2b_val_loss.png",
    title="2B initial convergence diagnostic",
    ylabel="Bounded validation loss",
)
```
